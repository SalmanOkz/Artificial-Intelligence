```python
import numpy as np
import pandas as pd


# LOAD DATASET

# Replace '2.csv' with your dataset file path
data = pd.read_csv('/content/2.csv')

# Features (all columns except last)
concepts = np.array(data.iloc[:, :-1])

# Target class (last column)
target = np.array(data.iloc[:, -1])


# CANDIDATE ELIMINATION


def candidate_elimination(concepts, target):

    # Initialize Specific Hypothesis
    specific_h = concepts[0].copy()

    # Initialize General Hypothesis
    general_h = [
        ['?' for _ in range(len(specific_h))]
        for _ in range(len(specific_h))
    ]

    print("Initial Specific Hypothesis:")
    print(specific_h)

    print("\nInitial General Hypothesis:")
    for g in general_h:
        print(g)

    print("\n=============================\n")

    # Process each training instance
    for i, instance in enumerate(concepts):

        print(f"Training Instance {i + 1}")
        print("Instance :", instance)
        print("Target   :", target[i])


        # POSITIVE EXAMPLE

        if target[i].lower() == "yes":

            print("\nPositive Example")

            for j in range(len(specific_h)):

                # Generalize specific hypothesis
                if instance[j] != specific_h[j]:
                    specific_h[j] = '?'
                    general_h[j][j] = '?'


        # NEGATIVE EXAMPLE

        elif target[i].lower() == "no":

            print("\nNegative Example")

            for j in range(len(specific_h)):

                # Specialize general hypothesis
                if instance[j] != specific_h[j]:
                    general_h[j][j] = specific_h[j]

                else:
                    general_h[j][j] = '?'


        # DISPLAY CURRENT STATE


        print("\nUpdated Specific Hypothesis:")
        print(specific_h)

        print("\nUpdated General Hypothesis:")
        for g in general_h:
            print(g)

        print("\n-----------------------------\n")

    # Remove overly general hypotheses
    final_general_h = []

    for g in general_h:
        if g != ['?' for _ in range(len(specific_h))]:
            final_general_h.append(g)

    return specific_h, final_general_h



# RUN ALGORITHM


specific_h, general_h = candidate_elimination(concepts, target)


# FINAL OUTPUT


print("\n=============================")
print("FINAL HYPOTHESIS")
print("=============================\n")

print("Final Specific Hypothesis:")
print(specific_h)

print("\nFinal General Hypothesis:")

for g in general_h:
    print(g)
```
