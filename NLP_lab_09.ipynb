```python
import pandas as pd
import re
import string
from textblob import TextBlob
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
import emoji
import nltk

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
data = {
    "review": [
        "I LOVE this product!!! 😍 Visit https://example.com",
        "<p>This is AMAZING!!!</p>",
        "LOL this is gr8 😂😂",
        "ceertain conditionas duriing seveal ggenerations."
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
df['review'] = df['review'].str.lower()
```

```python
def remove_html(text):
    return re.sub('<.*?>', '', text)

df['review'] = df['review'].apply(remove_html)
```

```python
def remove_url(text):
    return re.sub(r'https?://\S+|www\.\S+', '', text)

df['review'] = df['review'].apply(remove_url)
```

```python
def remove_punc(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df['review'] = df['review'].apply(remove_punc)
```

```python
chat_words = {
    "LOL": "Laughing Out Loud",
    "GR8": "Great"
}

def chat_convert(text):
    words = []
    for w in text.split():
        if w.upper() in chat_words:
            words.append(chat_words[w.upper()])
        else:
            words.append(w)
    return " ".join(words)

df['review'] = df['review'].apply(chat_convert)
```

```python
def correct_spelling(text):
    return str(TextBlob(text).correct())

df['review'] = df['review'].apply(correct_spelling)
```

```python
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])

df['review'] = df['review'].apply(remove_stopwords)
```

```python
def remove_emoji(text):
    return emoji.replace_emoji(text, replace='')

df['review'] = df['review'].apply(remove_emoji)
```

```python
df['review'] = df['review'].apply(lambda x: emoji.demojize(x))
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
df['tokens'] = df['review'].apply(word_tokenize)
print(df['tokens'])
```

```python
sentences = sent_tokenize("I live in Pakistan. I love Karachi.")
print(sentences)
```

```python
ps = PorterStemmer()

def stemming(text):
    return " ".join([ps.stem(word) for word in text.split()])

df['stemmed'] = df['review'].apply(stemming)
```

```python
lemmatizer = WordNetLemmatizer()

def lemmatization(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])

df['lemmatized'] = df['review'].apply(lemmatization)
```

```python
# ==============================
# 1. Import Libraries
# ==============================
import pandas as pd
import numpy as np
import re
import string

# NLP libraries
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required datasets
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# ==============================
# 2. Create Dataset (Movie Example)
# ==============================
data = {
    "Movie": ["Matrix", "3 Idiots"],
    "Description": [
        "A hacker discovers reality is simulated!!! 😮 Visit https://movie.com",
        "Story of 3 students :) www.film.com, awsome frndship!!!"
    ],
    "Genre": ["Action", "Comedy"]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)
```

```python
# ==============================
# 3. Lower Casing
# ==============================
df['clean'] = df['Description'].str.lower()
```

```python
# ==============================
# 4. Remove HTML Tags
# ==============================
def remove_html(text):
    pattern = re.compile('<.*?>')
    return pattern.sub('', text)

df['clean'] = df['clean'].apply(remove_html)
```

```python
# ==============================
# 5. Remove URLs
# ==============================
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub('', text)

df['clean'] = df['clean'].apply(remove_url)
```

```python
# ==============================
# 6. Remove Punctuation
# ==============================
def remove_punc(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df['clean'] = df['clean'].apply(remove_punc)
```

```python
# ==============================
# 7. Chat Word Treatment
# ==============================
chat_words = {
    "frnd": "friend",
    "awsome": "awesome",
    "luv": "love",
    "bcz": "because"
}

def chat_conversion(text):
    new_text = []
    for word in text.split():
        if word in chat_words:
            new_text.append(chat_words[word])
        else:
            new_text.append(word)
    return " ".join(new_text)

df['clean'] = df['clean'].apply(chat_conversion)
```

```python
# ==============================
# 8. Remove Stopwords
# ==============================
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])

df['clean'] = df['clean'].apply(remove_stopwords)
```

```python
# ==============================
# 9. Remove Emojis
# ==============================
def remove_emoji(text):
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F"
        u"\U0001F300-\U0001F5FF"
        u"\U0001F680-\U0001F6FF"
        u"\U0001F1E0-\U0001F1FF"
    "]+", flags=re.UNICODE)
    return emoji_pattern.sub('', text)

df['clean'] = df['clean'].apply(remove_emoji)
```

```python
# ==============================
# 10. Tokenization (Simple)
# ==============================
df['tokens'] = df['clean'].apply(lambda x: x.split())
```

```python
# ==============================
# 11. Stemming
# ==============================
ps = PorterStemmer()

def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])

df['stemmed'] = df['clean'].apply(stem_words)
```

```python
# ==============================
# 12. Lemmatization
# ==============================
lemmatizer = WordNetLemmatizer()

def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])

df['lemmatized'] = df['clean'].apply(lemmatize_words)
```

```python
# ==============================
# 13. Final Output
# ==============================
print("\nProcessed Data:\n")
print(df[['Description', 'clean', 'tokens', 'stemmed', 'lemmatized']])
```
